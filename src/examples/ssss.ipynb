```python
import json

# Load JSON data from a file
with open('cards_response.json', 'r') as file:
    data = json.load(file)

# print(data)
```

```python
def get_entry(item):
    entries = []
    entries.append({
        'label': item['name'],
        'url': item['iconUrls']['medium'],
    })
    if 'evolutionMedium' in item['iconUrls']:
        entries.append({
            'label': item['name'] + ' Evolution',
            'url': item['iconUrls']['evolutionMedium'],
        })
    return entries
```

```python
entries = []
for item in data['items']:
    for e in get_entry(item):
        entries.append(e)
for item in data['supportItems']:
    for e in get_entry(item):
        entries.append(e)
```
